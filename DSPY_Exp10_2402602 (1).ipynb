##1)Explain input and output characteristics for f_oneway function from stats library.
ANS:-

The **`f_oneway`** function from **`scipy.stats`** is used to perform **one-way ANOVA** in Python. It tests whether the means of **two or more independent groups** are significantly different. Let’s break down its **input and output characteristics**.


1)Input Characteristics:-

`scipy.stats.f_oneway` takes **two or more arrays (or lists) of numerical data**, each representing **a group**.

Required Inputs:-

1. **Group Data** – Each argument is a separate group:

   * Must be **numerical values** (int or float)
   * Can be a **list, tuple, or NumPy array**
   * All groups **must have at least 2 observations**

**Syntax:**

```python
f_oneway(group1, group2, group3, ...)
```

**Example Input:**

```python
group1 = [85, 90, 88, 75, 95]
group2 = [80, 78, 84, 82, 79]
group3 = [92, 94, 90, 96, 91]

stats.f_oneway(group1, group2, group3)
```

Notes on Input:

* Supports **2 or more groups** (ANOVA requires at least 2)
* Data should be **independent and approximately normally distributed**
* Variances should be roughly **homogeneous** (assumption for ANOVA)


2)Output Characteristics

`f_oneway` returns **two values** as a **tuple**:

```python
F-statistic, p-value
```

(a) F-statistic:-

* Measures the **ratio of between-group variance to within-group variance**.
* **High F-value** → More likely that group means are different.

(b) p-value:-

* Probability of observing the F-statistic (or more extreme) **if the null hypothesis is true**.
* Used to make the decision about **H₀: all group means are equal**.


3️)How to Interpret:-

| Output          | Meaning                                                        |
| --------------- | -------------------------------------------------------------- |
| **p-value < α** | Reject H₀ → At least one group mean is significantly different |
| **p-value ≥ α** | Do not reject H₀ → No significant difference among group means |

**Example Output:**

```python
F-statistic: 10.45
p-value: 0.0012
```

* Since **p < 0.05**, we reject the null hypothesis → significant difference exists.


##2)Explain results of sm.stats.anova_lm fucntion.
ANS:-

The **`sm.stats.anova_lm`** function from **statsmodels** is used to perform **ANOVA (Analysis of Variance)** in Python. It provides a **detailed table of variance analysis** that helps you decide whether group means are significantly different. Let’s go through its **results and how to interpret them**.


1️)Output of `anova_lm`:-

The function returns a **DataFrame (table)** with the following **key columns**:

| Column     | Description                                                                                                                                                                      |
| ---------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **sum_sq** | **Sum of squares** for each factor and residuals. It measures **total variation** attributed to that factor or error.                                                            |
| **df**     | **Degrees of freedom** associated with the factor or residuals. Calculated as **number of levels − 1** for a factor and **total observations − number of groups** for residuals. |
| **F**      | **F-statistic** for the factor. Ratio of **mean square of the factor / mean square of residuals**. Higher F → more likely the factor significantly affects the outcome.          |
| **PR(>F)** | **p-value** corresponding to the F-statistic. Probability of observing such an F if **null hypothesis is true**. Small p-value (< α) indicates significance.                     |


2️)Structure of the ANOVA Table:-

A typical one-way ANOVA table looks like this:

| Source   | sum_sq | df | F     | PR(>F) |
| -------- | ------ | -- | ----- | ------ |
| C(group) | 950.0  | 2  | 10.45 | 0.0012 |
| Residual | 500.0  | 12 | NaN   | NaN    |

**Explanation:**

1. **C(group)** – Variation **between the groups** (effect of the factor).
2. **Residual** – Variation **within groups** (random error).
3. **F = (MS_between / MS_within)** – Indicates whether between-group variance is larger than within-group variance.
4. **PR(>F) = p-value** – Used to test **H₀: all group means are equal**.


3️)How to Interpret Results:-

1. **Check the p-value (PR(>F))**:

   * **p < 0.05** → Reject H₀ → At least one group mean is different.
   * **p ≥ 0.05** → Do not reject H₀ → No significant difference.

2. **F-statistic**:

   * Large F → Strong evidence that group means differ.
   * Small F → Weak evidence, variation mostly due to random error.

3. **sum_sq and df**:

   * Used internally to compute **F-statistic**.
   * Can also help in **effect size calculations**.



4️)Example in Python:-

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Sample data:-
data = pd.DataFrame({
    'score': [85, 90, 88, 75, 95, 80, 78, 84, 82, 79, 92, 94, 90, 96, 91],
    'group': ['A']*5 + ['B']*5 + ['C']*5
})

# Fit model:-
model = ols('score ~ C(group)', data=data).fit()

# Perform ANOVA:-
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

**Output (example):**

| Source   | sum_sq | df | F     | PR(>F) |
| -------- | ------ | -- | ----- | ------ |
| C(group) | 950.0  | 2  | 10.45 | 0.0012 |
| Residual | 500.0  | 12 | NaN   | NaN    |

**Interpretation:**

* **p = 0.0012 < 0.05** → Reject H₀ → **group means are significantly different**.
* **F = 10.45** → Between-group variance is much higher than within-group variance.


##3)What role ANOVA plays in data analytics?
ANS:-

In **data analytics**, **ANOVA (Analysis of Variance)** plays a crucial role in **understanding patterns, relationships, and differences between groups**. It is widely used when analyzing **categorical factors and their impact on numerical outcomes**. Here’s a detailed breakdown of its role:


1️)Comparing Multiple Groups Efficiently:-

* ANOVA allows analysts to **compare the means of three or more groups at once**.
* Without ANOVA, we would need **multiple t-tests**, which increases the risk of **Type I error**.
* Example: Comparing **sales across 4 regions** to see if any region performs differently.


2️)Testing Hypotheses in Experiments:-

* Helps determine if an **independent variable (categorical)** has a **significant effect** on a **dependent variable (numeric)**.
* Common in **A/B testing, marketing experiments, or clinical trials**.
* Example: Checking whether **different ad campaigns** lead to significantly different **click-through rates**.


3️)Identifying Important Factors:-

* ANOVA highlights which **factors or variables have a statistically significant effect**.
* Analysts can focus on variables that **truly influence the outcome**.
* Example: In manufacturing, testing **different machine settings** to find which significantly affects **product quality**.

4️)Foundation for Post-Hoc Analysis:-

* When ANOVA indicates a significant difference, **post-hoc tests (e.g., Tukey HSD)** are used to identify **which specific groups differ**.
* This is valuable in analytics for **actionable insights**.


5️)Role in Regression and Predictive Modeling:-

* One-way ANOVA is conceptually linked to **categorical variable analysis in regression**.
* Helps in **feature selection**: a factor that doesn’t show a significant effect may be **less important for prediction**.


6️)Summarizing Variability in Data:-

* ANOVA decomposes the **total variability** into:

  * **Between-group variance** (variation due to factor)
  * **Within-group variance** (random variation)
* This helps analysts understand **sources of variation** in the data.


7️)Practical Examples in Data Analytics:-

1. **E-commerce**: Does **average purchase value** differ by **customer segment**?
2. **Healthcare**: Do **drug dosages** produce significantly different **blood pressure readings**?
3. **Education**: Do **different teaching methods** result in different **exam scores**?


##4)When to use one way ANOVA? Explain with real world case.
ANS:-

A **one-way ANOVA** (Analysis of Variance) is used when you want to **compare the means of three or more independent groups** to see if there is a statistically significant difference among them. The key points for using one-way ANOVA are:

1. **Single independent variable (factor)** – with two or more levels (groups).
2. **Dependent variable is continuous** – numeric data like weight, height, score, etc.
3. **Groups are independent** – no repeated measures on the same subjects.
4. **Normality & Homogeneity of variance** – the dependent variable should be roughly normally distributed within each group, and variances should be similar.

**Real-World Example**

**Scenario:** A company wants to test if different training programs affect employee productivity.

* **Independent variable (factor):** Training Program

  * Level 1: Online course
  * Level 2: In-person workshop
  * Level 3: Self-study manual
* **Dependent variable:** Productivity score (numeric, e.g., tasks completed per week)

**Goal:** Determine if the average productivity differs between the three training programs.

**How one-way ANOVA helps:**

* Instead of doing multiple t-tests (which increases the chance of Type I error), one-way ANOVA tests all three groups **simultaneously**.
* If ANOVA shows a significant difference (p < 0.05), we conclude that at least one group’s mean productivity is different.
* Post-hoc tests (like Tukey’s HSD) can then tell **which specific groups differ**


##5)When to use 2 way ANOVA ? Explain with real world case.
ANS:-

A **two-way ANOVA** (analysis of variance) is used when you want to examine the effect of **two independent categorical variables (factors)** on a **continuous dependent variable**, and also see if there is an **interaction effect** between the two factors.

Let’s break it down clearly.


**When to Use Two-Way ANOVA**

1. **Two independent factors:** You have two categorical variables (e.g., treatment type, gender, region, etc.).
2. **Continuous dependent variable:** The outcome you measure is numeric (e.g., blood pressure, test scores, sales amount).
3. **Interest in interaction:** You want to see if the effect of one factor depends on the level of the other factor.

* **Main effect:** Effect of one factor ignoring the other.
* **Interaction effect:** Combined effect of the two factors. Sometimes the effect of factor A changes depending on factor B.

**Real-World Case Example**

**Scenario:** A company wants to test a new training program for employees and sees if it improves productivity. They consider two factors:

1. **Type of Training:** Online vs. In-person
2. **Experience Level:** Junior vs. Senior employees

**Dependent variable:** Number of tasks completed in a week (continuous).

**Research Questions:**

1. Does the **type of training** affect productivity? (main effect of training)
2. Does **experience level** affect productivity? (main effect of experience)
3. Does the **effect of training depend on experience level**? (interaction effect)

**Data Layout Example**

| Training Type | Experience Level | Tasks Completed |
| ------------- | ---------------- | --------------- |
| Online        | Junior           | 35              |
| Online        | Junior           | 40              |
| Online        | Senior           | 50              |
| In-person     | Junior           | 45              |
| In-person     | Senior           | 55              |

Here, a **two-way ANOVA** will tell you:

* Whether training type has a significant effect on productivity.
* Whether experience level has a significant effect.
* Whether the combination (interaction) of training type and experience level significantly affects productivity.
