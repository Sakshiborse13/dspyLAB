##1)What is 2 tailed and  tailed T Test ?
ANS:-

In **statistics**, a **t-test** is used to check whether a sample mean is significantly different from a hypothesized value or another sample mean. There are two common forms based on the **direction of the hypothesis**: **one-tailed** and **two-tailed** tests.

1)One-Tailed t-Test:-

A **one-tailed t-test** checks if the mean is **either greater than OR less than** a specific value (only one direction).

Idea:-

You test **one side of the distribution**.

Example:-

Suppose a company claims the average battery life of a phone is **10 hours**.

You want to test if the battery life is **greater than 10 hours**.

* **Null hypothesis (H₀):** μ = 10
* **Alternative hypothesis (H₁):** μ > 10

Here you only care about **higher values**, so it's **one-tailed (right-tailed)**.

Types of One-Tailed Tests:-

1. **Right-tailed test** → H₁: μ > μ₀
2. **Left-tailed test** → H₁: μ < μ₀

Graphically:-

Only **one tail of the distribution** is used to determine the rejection region.


2)Two-Tailed t-Test:-

A **two-tailed t-test** checks if the mean is **different (either higher or lower)** from a specific value.

Idea:-

You test **both sides of the distribution**.

Example:-

Battery life claim = **10 hours**, but you want to check if it is **different from 10 hours**.

* **Null hypothesis (H₀):** μ = 10
* **Alternative hypothesis (H₁):** μ ≠ 10

Here the mean could be **higher OR lower**, so both tails are considered.

Graphically:-

The rejection regions are **split into two tails**.

If α = 0.05

* Left tail = 0.025
* Right tail = 0.025



##2)Explain what is Type I and Type II errors with an example.
ANS:-


1)Type I Error (False Positive):-

A **Type I error** happens when we **reject the null hypothesis even though it is actually true**.

* Symbol: **α (alpha)**
* Also called **False Positive**

Example:-

Suppose a **medical test** checks whether a person has a disease.

* **H₀:** The person **does not have the disease**
* **H₁:** The person **has the disease**

If the test result says the person **has the disease**, but in reality they **do not**, then this is a **Type I error**.



2)Type II Error (False Negative):-

A **Type II error** happens when we **fail to reject the null hypothesis even though it is false**.

* Symbol: **β (beta)**
* Also called **False Negative**

Example:-

Using the same medical test:

* **H₀:** The person **does not have the disease**
* **H₁:** The person **has the disease**

If the test says the person **does not have the disease**, but actually **they do**, this is a **Type II error**.




##3)What is power of statistical test?
ANS:-

**Power of a Statistical Test** is the **probability that the test correctly rejects a false null hypothesis (H₀)**.

In simple words, it measures **how good the test is at detecting a real effect when it actually exists**.

Mathematical Definition:-

[
\text{Power} = 1 - \beta
]

Where:

* **β (beta)** = probability of **Type II error**
* **Power** = probability of **not making a Type II error**


Explanation:-

When we perform hypothesis testing:

* **H₀:** Null hypothesis
* **H₁:** Alternative hypothesis

If **H₀ is false**, a good statistical test should **reject it**.
The **power** tells us how likely the test is to do that correctly.


Example:-

Suppose a researcher is testing a **new medicine**.

* **H₀:** The medicine has **no effect**
* **H₁:** The medicine **works**

If the medicine actually works and the test **detects this effect**, the test has **high power**.

If the test fails to detect the effect, that is a **Type II error**.

So:

[
\text{Power} = 1 - \text{Probability of Type II error}
]


Interpretation:-

* **High Power (e.g., 0.8 or 80%)** → Test is good at detecting real effects
* **Low Power** → Test may miss real differences

Researchers often aim for **80% or 90% power** in studies.


Factors That Affect Power:-

1. **Sample size** – Larger sample → Higher power
2. **Significance level (α)** – Higher α → Higher power
3. **Effect size** – Larger effect → Higher power
4. **Variability in data** – Lower variability → Higher power



##4)Explain input and out put characteristics for  stats.ttest_rel function in your own words.
ANS:-

1)Input Characteristics of `stats.ttest_rel()`:-

The inputs are the **data values that the function needs to perform the test**.

Main Inputs:-

1. **First Sample (a)**

   * A list or array of numerical values.
   * Represents the **first set of related observations**.

2. **Second Sample (b)**

   * Another list or array of numbers.
   * Represents the **second set of observations corresponding to the first sample**.

3. **Equal Length Requirement**

   * Both samples must have the **same number of observations** because each value in one sample is paired with the corresponding value in the other sample.

4. **Optional Parameters**

   * **axis** – Specifies the axis along which the test is performed.
   * **nan_policy** – Specifies how to handle missing values (ignore, propagate, etc.).
   * **alternative** – Defines the type of test:

     * `"two-sided"`
     * `"less"`
     * `"greater"`

Example Input:-

* Marks of students **before training**
* Marks of the **same students after training**

Each student contributes **two related values**, forming a pair.



2)Output Characteristics of `stats.ttest_rel()`:-

The function returns the **results of the statistical test**.

Main Outputs:-

1. **t-statistic**

   * Measures the **difference between the paired sample means relative to the variation in the data**.
   * Larger absolute values indicate a **stronger difference** between the samples.

2. **p-value**

   * Shows the **probability of obtaining the observed result if the null hypothesis is true**.
   * Used to decide whether the difference is **statistically significant**.

Decision Rule:-

* If **p-value < significance level (α, e.g., 0.05)** → Reject the null hypothesis.
* If **p-value ≥ α** → Do not reject the null hypothesis.


##5)What is function in stats module for identifying power of a statistical test. Explain out put range of it and its significance.
ANS:-


 `statsmodels.stats.power.ttest_power()`:-

This function calculates the **power of a t-test**, which tells us the **probability of correctly rejecting the null hypothesis when it is false**.


1) Purpose of the Function:-

`ttest_power()` is used to determine **how likely a t-test is to detect a real difference or effect** in the data.

Researchers use it during **experiment design** to check whether their **sample size is sufficient** to detect an effect.



2)Input Parameters (Main Ones):-

1. **effect_size**

   * The magnitude of the difference between the population mean and the null hypothesis value.

2. **nobs**

   * Number of observations (sample size).

3. **alpha**

   * Significance level (commonly **0.05**).

4. **alternative**

   * Type of test:
   * `"two-sided"`
   * `"larger"`
   * `"smaller"`

3)Output of the Function:-

The function returns **a single value representing the statistical power of the test**.

Output Range:-

[
0 \leq \text{Power} \leq 1
]

* **0** → Test has **no ability** to detect the effect.
* **1** → Test will **always detect the effect** when it exists.

In practice:

* **0.80 (80%) power** is commonly considered **adequate** in research.



4)Significance of the Output:-

The output value tells us **how reliable the statistical test is in detecting real differences**.

* **High power (≥ 0.8)**
  → Low probability of **Type II error (β)**
  → Test is good at detecting real effects.

* **Low power (< 0.5)**
  → High chance of **missing a real effect**.

Since:

[
\text{Power} = 1 - \beta
]

where **β is the probability of a Type II error**.
