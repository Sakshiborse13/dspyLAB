##1)When do we go for Chi square test ?
ANS:-

We use the **Chi-Square (χ²) test** when we want to analyze **categorical data** and check whether there is a **significant difference between observed and expected frequencies**.

The Chi-Square test is commonly used in statistics and is available in libraries such as **SciPy** for data analysis.


When Do We Use a Chi-Square Test?

1️)When Data Is Categorical:-

The variables should be **categories (counts or frequencies)** rather than numerical measurements.

**Example:**

* Gender (Male / Female)
* Product preference (A / B / C)


2️)To Test Independence Between Two Variables:-

Used to determine whether **two categorical variables are related** or **independent**.

**Example:**
Checking whether **gender and product preference are related**.

* H₀: Variables are **independent**
* H₁: Variables are **dependent**


3️)To Test Goodness of Fit:-

Used to check whether **observed frequencies match expected frequencies** according to a theoretical distribution.

**Example:**
Testing whether a **dice is fair**.

* Expected frequency for each number = same
* Compare with observed results.

4️)When Sample Data Is in Frequency Form:-

The test works with **counts (number of occurrences)**.

Example table:

| Category | Observed Frequency |
| -------- | ------------------ |
| A        | 20                 |
| B        | 15                 |

##2)Do we use Chi square test for independence ? Explain How ?
ANS:-

Yes, the **Chi-Square (χ²) test** is commonly used for **testing independence between two categorical variables**. It helps determine whether the variables are **related (dependent)** or **not related (independent)**.


1️)Purpose of Chi-Square Test for Independence:-

The test checks whether **two categorical variables influence each other**.

Hypotheses:-

* **Null Hypothesis (H₀):** The two variables are **independent**.
* **Alternative Hypothesis (H₁):** The two variables are **dependent (associated)**.



2️)How the Test Is Performed:-

Step 1: Collect Data in a Contingency Table:-

Data is arranged in a **frequency table**.

Example: Relationship between **Gender** and **Product Preference**

| Gender | Product A | Product B | Total |
| ------ | --------- | --------- | ----- |
| Male   | 30        | 20        | 50    |
| Female | 20        | 30        | 50    |
| Total  | 50        | 50        | 100   |



step 2: Calculate Expected Frequencies:-

Expected value for each cell:

[
E = \frac{(\text{Row Total} \times \text{Column Total})}{\text{Grand Total}}
]

Example for **Male–Product A**:

[
E = \frac{50 \times 50}{100} = 25
]



Step 3: Compute Chi-Square Statistic:-

[
\chi^2 = \sum \frac{(O - E)^2}{E}
]

Where:

* **O** = Observed frequency
* **E** = Expected frequency

Step 4: Determine Degrees of Freedom:-

[
df = (r-1)(c-1)
]

Where

* **r** = number of rows
* **c** = number of columns

Step 5: Compare with Critical Value or p-value

* If **χ² calculated > χ² table value** (or **p < α**), reject **H₀**.
* Otherwise, do not reject **H₀**.

3️)Interpretation:-

* **Reject H₀:** Variables are **dependent (associated)**.
* **Do not reject H₀:** Variables are **independent**.


##3)Explain role of Critical value and p-value in Chi square test.
ANS:-

In a **Chi-Square (χ²) test**, both **critical value** and **p-value** help us decide whether to **accept or reject the null hypothesis (H₀)**. They are two different methods used to interpret the result of the test.

1️)Role of Critical Value in Chi-Square Test:-

The **critical value** is the **threshold value** taken from the **Chi-Square distribution table** based on:

* **Significance level (α)** (commonly 0.05)
* **Degrees of freedom (df)**

How It Is Used:-

1. Calculate the **χ² test statistic** from the data.
2. Find the **critical value** from the Chi-Square table.
3. Compare the two values.

Decision Rule:-

* If **χ² calculated > χ² critical** → **Reject H₀**
* If **χ² calculated ≤ χ² critical** → **Do not reject H₀**

Meaning:-

The critical value defines the **boundary of the rejection region** for the test.


2️)Role of p-Value in Chi-Square Test:-

The **p-value** represents the **probability of obtaining the observed result (or more extreme) assuming the null hypothesis is true**.

How It Is Used:-

1. Calculate the **p-value** from the χ² statistic.
2. Compare it with the **significance level (α)**.

Decision Rule:-

* If **p-value < α (e.g., 0.05)** → **Reject H₀**
* If **p-value ≥ α** → **Do not reject H₀**

Meaning:-

The p-value tells us **how strong the evidence is against the null hypothesis**.
   


##4)Can we use chi square test to show existence of correlation between 2 features in a dataset ?  How will you interpret test result to indicate correlation presence.
ANS:-

Yes, the **Chi-Square (χ²) test** can be used to check whether there is a **relationship (association) between two categorical features in a dataset**. This is called the **Chi-Square Test of Independence**, and it is commonly implemented in statistical libraries like **SciPy**.

However, note that the Chi-Square test **does not measure the strength of correlation** like Pearson correlation does. It only tells us **whether an association exists or not**.

1️)When Chi-Square Indicates Correlation:-

We use it when **both features are categorical**.

**Example features**

* Gender (Male / Female)
* Purchase decision (Yes / No)

We want to check whether **gender influences purchase decision**.

2)Hypotheses:-

* **Null Hypothesis (H₀):** The two features are **independent** (no correlation/association).
* **Alternative Hypothesis (H₁):** The two features are **dependent** (there is correlation/association).

3️)Steps to Perform the Test:-

Step 1: Create a Contingency Table

| Gender | Purchase Yes | Purchase No |
| ------ | ------------ | ----------- |
| Male   | 40           | 20          |
| Female | 30           | 10          |



Step 2: Calculate Expected Frequencies:-

[
E = \frac{(\text{Row Total} \times \text{Column Total})}{\text{Grand Total}}
]

These expected values assume the variables are **independent**.

Step 3: Calculate Chi-Square Statistic:-

[
\chi^2 = \sum \frac{(O - E)^2}{E}
]

Where

* **O** = observed frequency
* **E** = expected frequency

4️)Interpreting the Result:-

We interpret the result using **p-value or critical value**.

Using p-value:-

* **p-value < α (e.g., 0.05)**
  → Reject **H₀**
  → The features are **associated (correlated)**

* **p-value ≥ α**
  → Do not reject **H₀**
  → No evidence of correlation (features are independent)

Using Critical Value:-

* **χ² calculated > χ² critical** → Association exists
* **χ² calculated ≤ χ² critical** → No association

5️)inal Interpretation:-

If the **null hypothesis is rejected**, it means:

* The two features are **not independent**
* There is **statistical evidence of association (correlation)** between them.

If **H₀ is not rejected**, it means:

* The features are **independent**
* No significant correlation is detected.


##5)What is degrees of freedom and how to calculate it while performing chi square test.
ANS:-

In the context of the **Chi-Square (χ²) test**, **degrees of freedom (df)** is a measure of the **number of independent values or pieces of information that can vary** when estimating statistical parameters. It’s a key component because it determines the **shape of the χ² distribution**, which is used to find critical values or p-values.

1)Definition:-

* **Degrees of freedom (df)** = Number of values that are **free to vary** in the calculation of a statistic.
* It affects the **critical value** of the Chi-Square test: higher df → wider χ² distribution.

**Intuition:** Once you know the totals in a contingency table, not all values can vary freely. Some are fixed by the row and column totals.


2)How to Calculate Degrees of Freedom in Chi-Square Test:-

(a) For Test of Independence

For a **contingency table**:

[
\text{df} = (\text{number of rows} - 1) \times (\text{number of columns} - 1)
]

**Example:** 2×3 table (2 rows, 3 columns)

[
\text{df} = (2-1) \times (3-1) = 1 \times 2 = 2
]



(b) For Goodness-of-Fit Test

For a **single categorical variable**:

[
\text{df} = (\text{number of categories}) - 1
]

**Example:** Rolling a dice with 6 sides

[
\text{df} = 6 - 1 = 5
]



3️)Why Degrees of Freedom Matter:-

* Determines the **critical value** from the χ² distribution table.
* Helps adjust for the **number of categories or constraints**, so we don’t overestimate significance.
* Used to calculate **p-value** for hypothesis testing.
